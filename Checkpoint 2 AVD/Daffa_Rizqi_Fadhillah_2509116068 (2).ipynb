#**Outliers**

```python
results = []
cols = df.select_dtypes(include=['float64', 'int64'])

for col in cols:
    q1 = df[col].quantile(0.25)
    q3 = df[col].quantile(0.75)
    iqr = q3 - q1
    lower_bound = q1 - 1.5*iqr
    upper_bound = q3 + 1.5*iqr
    outliers = df[(df[col] < lower_bound) | (df[col] > upper_bound)]
    percent_outliers = (len(outliers)/len(df))*100
    results.append({'Kolom': col, 'Persentase Outliers': percent_outliers})

pd.DataFrame(results)
```

Berdasarkan hasil pengecekan, ditemukan beberapa data dengan simbol pada nilai yang menyebabkan sistem tidak dapat membaca data tersebut. Namun, kolom-kolom tersebut akan serga diperbaiki agar outliers dapat membaca sistem dan menampilkan output.

Sementara itu, kategori seperti Office Products, Musical Instruments, Toys & Games, Car & Motorbike, dan Health & Personal Care memiliki jumlah produk yang jauh lebih sedikit.
Hal ini menunjukkan bahwa dataset didominasi oleh produk-produk elektronik dan perlengkapan teknologi, sehingga analisis yang dilakukan kemungkinan besar akan lebih merepresentasikan karakteristik pasar produk elektronik dibandingkan kategori lainnya.

##**Compoition**

```python
# Clean and convert 'rating_count' to numeric
df['rating_count'] = df['rating_count'].str.replace(',', '', regex=True) # Remove commas
df['rating_count'] = pd.to_numeric(df['rating_count'], errors='coerce') # Convert to numeric, coerce errors to NaN
df['rating_count'] = df['rating_count'].fillna(0).astype(int) # Fill NaN with 0 and convert to integer

sales_by_category = df.groupby('main_category')['rating_count'].sum().sort_values(ascending=False)

top5 = sales_by_category.head(5)

plt.figure(figsize=(8,8))
plt.pie(top5.values, labels=top5.index, autopct='%1.1f%%')
plt.title('Komposisi Total Rating Count per Main Category (Top 5)')
plt.show()
```

Berdasarkan pie chart komposisi total rating count per main category, terlihat bahwa kategori Electronics mendominasi dengan kontribusi sekitar 59% dari total rating. Disusul oleh Computers & Accessories sebesar kurang lebih 28,9%, dan Home & Kitchen sekitar 11,2%. Sementara kategori lainnya seperti Office Products dan Musical Instruments hanya memberikan kontribusi yang sangat kecil.
Hal ini menunjukkan bahwa sebagian besar interaksi dan minat pelanggan terpusat pada produk elektronik dan perlengkapan komputer. Distribusi ini juga menandakan bahwa pasar dalam dataset sangat terkonsentrasi pada kategori teknologi dibandingkan kategori lainnya.

##**Relationship**

```python
plt.figure(figsize=(8, 6))
sns.heatmap(df[['actual_price', 'rating_count']].corr(),
            annot=True,
            cmap='viridis',
            fmt='.2f')
plt.title('Korelasi antara Harga dan Jumlah Rating')
plt.show()
```

Berdasarkan heatmap korelasi antara actual_price dan rating_count, diperoleh nilai korelasi sebesar -0.04. Nilai ini sangat mendekati nol, sehingga menunjukkan bahwa hubungan antara harga produk dan jumlah rating sangat lemah atau hampir tidak ada hubungan linear.
Artinya, produk dengan harga lebih mahal tidak selalu memiliki jumlah rating yang lebih banyak, dan produk dengan harga lebih murah juga tidak selalu memiliki jumlah rating yang lebih sedikit. Dengan kata lain, tingkat popularitas (diukur dari jumlah rating) tidak dipengaruhi secara signifikan oleh harga produk.
